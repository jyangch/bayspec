```python
from bayspec import Data, DataUnit, Pair
from bayspec.model.local import cpl
```

```python
model = cpl()
model.par[1].frozen_at(-1.6)
model.par[2].frozen_at(2.4)
model
```

```python
me = DataUnit(
    src='./ME/me.src',
    bkg='./ME/me.bkg',
    rsp='./ME/me.rsp',
    notc=[10, 1000],
    stat='pgstat',
    grpg={'min_sigma': 3, 'max_bin': 10},
)

data = Data([('ME', me)])
data
```

```python
pair = Pair(data, model)
ecf = pair.conv_rate_to_flux[0]
ecf
```

```python
ctsrate_limit = data.net_ctsrate_significance_limit(3)[0]
ctsrate_limit
```

```python
ecf * ctsrate_limit
```
